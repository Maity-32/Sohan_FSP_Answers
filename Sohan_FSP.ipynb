```python
# All the data frames were collected from my previous projects: the soil nutrition measurement device, line-following robot,water purifier project, Weather Check.
#When I built the soil nutrition measurement device, sometimes it missed capturing data.
# -----Sohan Maity
# Q.1
import pandas as pd
import numpy as np

data = {
    'Nitrogen_mg': [45.2, np.nan, 42.1, 47.5, np.nan],
    'Phosphorus_mg': [12.1, 15.2, np.nan, 14.5, 11.2],
    'Potassium_mg': [110.0, 115.0, 108.0, np.nan, 105.0]
}
df_soil = pd.DataFrame(data)

print("--- Missing Values Detected ---")
print(df_soil.isnull().sum())


df_filled = df_soil.copy()
df_filled['Nitrogen_mg'].fillna(df_filled['Nitrogen_mg'].mean(), inplace=True)
df_filled['Phosphorus_mg'].fillna(df_filled['Phosphorus_mg'].mean(), inplace=True)
df_filled['Potassium_mg'].fillna(df_filled['Potassium_mg'].median(), inplace=True)

print("\n--- After Replacing Missing Values ---")
print(df_filled)

df_dropped = df_soil.dropna()
print("\n--- After Dropping Rows with Missing Values ---")
print(df_dropped)
```

```python
 # Q.2
import pandas as pd

lfr_data = {
    'Timestamp_ms': [100, 120, 120, 140, 160, 160, 180],
    'Left_Array_Sum': [3, 2, 2, 0, 5, 5, 1],
    'Right_Array_Sum': [0, 1, 1, 0, 2, 2, 4]
}
df_lfr = pd.DataFrame(lfr_data)

print("--- Duplicate Rows ---")
duplicates = df_lfr[df_lfr.duplicated()]
print(duplicates)

df_cleaned = df_lfr.drop_duplicates()
print("\n--- Dataset After Removing Duplicates ---")
print(df_cleaned)
```

```python
# Q.3
values = [12, 25, 40, 55, 70]

x_min = min(values)
x_max = max(values)

normalized_values = [(x - x_min) / (x_max - x_min) for x in values]

print(f"Original values: {values}")
print(f"Normalized values: {normalized_values}")
```

```python
# Q.4
import pandas as pd
import numpy as np

turbidity_data = pd.DataFrame({'Turbidity_NTU': [4.5, 5.2, 4.8, 15.0, 4.9, 5.1]})

mean_val = turbidity_data['Turbidity_NTU'].mean()
std_dev = turbidity_data['Turbidity_NTU'].std(ddof=0) # ddof=0 for population std dev

turbidity_data['Z_Score'] = (turbidity_data['Turbidity_NTU'] - mean_val) / std_dev

print("--- Z-Score Normalization ---")
print(turbidity_data)
```

```python
# Q.5
import pandas as pd
import numpy as np

# Simulated ESP32 analog pin voltage readings (mV)
df_voltage = pd.DataFrame({'Voltage_mV': [3200, 3210, 3195, 3205, 4500, 3180, 1200]})

mean_v = df_voltage['Voltage_mV'].mean()
std_v = df_voltage['Voltage_mV'].std()
df_voltage['Z_Score'] = (df_voltage['Voltage_mV'] - mean_v) / std_v

threshold = 1.5 
outliers = df_voltage[np.abs(df_voltage['Z_Score']) > threshold]

print("--- Outliers Detected ---")
print(outliers)
```

```python
# Q.8
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Raw ADC values from sensors
df_adc = pd.DataFrame({
    'Sensor_A_ADC': [1023, 2048, 4095, 512, 1024],
    'Sensor_B_ADC': [255, 128, 64, 250, 100]
})

scaler = StandardScaler()

scaled_features = scaler.fit_transform(df_adc)

df_scaled = pd.DataFrame(scaled_features, columns=df_adc.columns)

print("--- Standard Scaled Data ---")
print(df_scaled)
```

```python
# Q.9
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Simulated DHT22 Environmental Data
df_env = pd.DataFrame({
    'Temp_C': [22.5, 25.0, 24.1, 35.0, 21.0],
    'Humidity_Pct': [45, 50, 48, 90, 42]
})

minmax_scaler = MinMaxScaler()
std_scaler = StandardScaler()

minmax_scaled = minmax_scaler.fit_transform(df_env)
df_minmax = pd.DataFrame(minmax_scaled, columns=['Temp_MinMax', 'Humidity_MinMax'])

std_scaled = std_scaler.fit_transform(df_env)
df_std = pd.DataFrame(std_scaled, columns=['Temp_Std', 'Humidity_Std'])

df_comparison = pd.concat([df_env, df_minmax, df_std], axis=1)

print("--- Comparison: Original vs Min-Max vs Standardization ---")
print(df_comparison)
```
